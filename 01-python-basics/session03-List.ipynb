```python
# List is a collection of values or items of different types 
#list1 = [value1, value2, ...]
l0=[]     #empty list
l1 = [123, "python" ,3.7 ]
l2 = [1,2,3,4,5,6 ]
l3 =[ "Python", "java programming" , "cpp"]
print (l0)
print (l1)
print (l2)
print (l3)

```

```python
#list indexing  (forward direction  and bckward direction)
l1 = [123, "python" ,3.7 ]
l2 = [1,2,3,4,5,6 ]
l3 =[ "Python", "java programming" , "cpp"]
print (l1[2])
print (l2[2:4])   #  [start:end:step]
print (l3[-1])
```

```python
# list operation (  +  * []  [:]   in not in )
l1 = [123, "python" ,3.7 ]
l2 = [1,2,3,4,5,6 ]
l3 =[ "Python", "java programming" , "cpp"]
print (l1 + l2 + l3 )
print (l2 *4)
print (5 in l2) 
print ("cpp" not in l3)
```

```python
#sorted(list)  use to sort all items of list in an ascending order (descending by reverse =True) it makes new list 
num =[10,2,3,4,5]
lang = ["java", "cpp", "php", "python"]
print(sorted(num,reverse=True))
print(sorted(lang))
result1 =sorted(lang ,key=len, reverse = True)
#result1 =sorted(num ,key=len, reverse = True)
print(result1)
print(sorted(lang, reverse = True))
```

```python
#sort(list)  use to sort the list elements (descending by reverse =True) it modifies the original list
num =[1,2,3,4,5]
lang = ["java", "cpp", "php", "python"]
num.sort( reverse = True)
lang.sort( reverse = True)
lang.sort(key=len, reverse = True)
print(num)
print(lang)
num.sort()
print(num)
lang.sort()
print(lang)
```

```python
#list.reverse()  it reverses elements of the list 
word = ['p','y','t','h','o', 'n']
print(word)
word.reverse()
print('after reversing')
print(word)
result = word.reverse()
print(result)
```

```python
#list.count() returns the number of specific item appears in the list, if the element is not present in the list, it returns 0
num=[1,2,3,4,5,6,7,2,5,2,2,8,9,0]
cnt =num.count(2)       
print("count of 2 is:",cnt)
cnt =num.count(5)
print("count of 5 is:",cnt)
cnt =num.count(10)  
print("count of 10 is:",cnt) 



a = [1, True]  
print("count of 1 is:", a.count(1)) 
```

```python
#list.index() it takes an argument and returns index of it. if the element is not present, it raises a ValueError

lang =['p','y','t','h','o','n','p','r','o','g','r','a','m']

print("index of 't'  is:",lang.index('t'))     #first 't' in the list
print("index of 'p'  is:",lang.index('p')) 

#print("index of 'z'  is:",lang.index('z'))    # we dont have 'z' so there is a ValuError 
print("index of 'p'  is:",lang.index('p',3,10))# between 3 :9  index(value, start ,end)  


a = [2, True, 1]   
print(a.index(1)) 
```
