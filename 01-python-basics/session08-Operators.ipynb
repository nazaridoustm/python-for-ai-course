# Operators in Python
An operator is a symbol that performs an operation on one or more values (operands).
in `a + b` :  `+` is the operator and `a` , `b` are the operands.

Python operators are classified into these groups :

    1. Arithmetic operators       + - * / // % **
    2. Assignment operators       = += -= *= /= //= %= **= :=
    3. Comparison operators       == != > < >= <=
    4. Logical operators          and or not
    5. Identity operators         is   is not
    6. Membership operators       in   not in
    7. Bitwise operators          & | ^ ~ << >>
    8. Matrix multiplication      @   (used by NumPy / PyTorch)

## 1. Arithmetic Operators

```python
# + - * / // % **
a = 17
b = 5
print("a + b  =", a + b)      # addition
print("a - b  =", a - b)      # subtraction
print("a * b  =", a * b)      # multiplication
print("a / b  =", a / b)      # true division  ---> ALWAYS float
print("a // b =", a // b)     # floor division ---> rounds DOWN
print("a % b  =", a % b)      # modulus (remainder)
print("a ** b =", a ** b)     # power
```

```python
# / always returns float , even when the result is a whole number
print(10 / 2)
print(type(10 / 2))
print(type(10 // 2))
print(10 // 2.0)              # if one operand is float , the result is float
```

```python
# divmod(a, b) returns (a // b , a % b) in one step
q, r = divmod(17, 5)          # unpacking
print(q, r)

# the golden rule :  a == (a // b) * b + (a % b)
a, b = 17, 5
print(a == (a // b) * b + (a % b))
```

```python
# // and % with NEGATIVE numbers   (Python floors towards minus infinity)
print(-7 // 2)        # -4   (not -3 !)
print(-7 % 2)         #  1   (the sign of the result follows the divisor)
print(7 % -2)         # -1
print(int(-7 / 2))    # -3   int() truncates towards zero

import math
print(math.fmod(-7, 2))   # -1.0  C-style remainder (sign follows the dividend)
```

```python
# ** power
print(2 ** 10)
print(2 ** -1)        # negative power ---> float
print(9 ** 0.5)       # square root
print(2 ** 3 ** 2)    # ** is RIGHT associative : 2 ** (3 ** 2) = 2 ** 9
print(-2 ** 2)        # -4  because ** is stronger than unary minus : -(2 ** 2)
print((-2) ** 2)      #  4
print(pow(2, 10, 1000))   # pow(base, exp, mod) = (2**10) % 1000 , fast for cryptography
```

```python
# arithmetic operators on other types (operator overloading)
print("ab" + "cd")        # concatenation
print("ab" * 3)           # repetition
print([1, 2] + [3])       # list concatenation
print([0] * 4)
print((1, 2) * 2)
#print("ab" - "b")        # TypeError: unsupported operand type(s) for -: 'str' and 'str'
#print("5" + 5)           # TypeError: can only concatenate str (not "int") to str
```

```python
# Application 1 : modulus %
print([n for n in range(10) if n % 2 == 0])     # even numbers

seconds = 7384
h, rem = divmod(seconds, 3600)
m, s = divmod(rem, 60)
print(f"{seconds} seconds = {h:02d}:{m:02d}:{s:02d}")

# log every 100 steps during training
for step in range(1, 501):
    if step % 100 == 0:
        print("step", step, "-> log the loss")
```

```python
# Application 2 : // and %  for batches
n_samples = 1050
batch_size = 32
full_batches = n_samples // batch_size
last_batch = n_samples % batch_size
n_batches = -(-n_samples // batch_size)       # ceiling division trick
print("full batches :", full_batches)
print("last batch   :", last_batch)
print("total batches:", n_batches, "=", math.ceil(n_samples / batch_size))
```

```python
# Application 3 : % for a circular (ring) index , e.g. 24-hour clock or ring buffer
hours = [22, 23, 0, 1]
print([(h + 5) % 24 for h in hours])

buffer_size = 4
for i in range(10):
    print(i % buffer_size, end=" ")
```

```python
# Application 4 : formulas in machine learning
x = [2.0, 4.0, 6.0, 8.0]
mean = sum(x) / len(x)
var = sum((xi - mean) ** 2 for xi in x) / len(x)
std = var ** 0.5
z = [(xi - mean) / std for xi in x]           # standardization (z-score)
print("mean =", mean, " std =", round(std, 4))
print("z-scores:", [round(v, 3) for v in z])

import math
def sigmoid(t):
    return 1 / (1 + math.e ** (-t))           # logistic function
print("sigmoid(0) =", sigmoid(0), " sigmoid(4) =", round(sigmoid(4), 4))
```

## 2. Assignment Operators

```python
x = 10
x += 5     # x = x + 5
print(x)
x -= 3     # x = x - 3
print(x)
x *= 2     # x = x * 2
print(x)
x /= 4     # x = x / 4   ---> float
print(x)
x //= 2    # x = x // 2
print(x)
x **= 3
print(x)
x %= 5
print(x)
```

```python
# += on a list changes the SAME object (in place) , + creates a NEW object
a = [1, 2]
b = a
a += [3]
print(a, b, a is b)

a = [1, 2]
b = a
a = a + [3]
print(a, b, a is b)

# += on immutable objects (int , str , tuple) always creates a new object
t = (1, 2)
old_id = id(t)
t += (3,)
print(t, id(t) == old_id)
```

```python
# Python has NO ++ and -- operators
i = 5
++i            # this is  +(+i)  ---> does nothing !
print(i)
i += 1         # the correct way
print(i)
```

```python
# := walrus operator (Python 3.8+) : assigns AND returns the value inside an expression
data = [3, 8, 1, 9, 4]
if (n := len(data)) > 3:
    print(f"list is long ({n} items)")

# useful to avoid computing something twice
values = [12, 7, 25, 3, 18]
scaled = [y for v in values if (y := v * 10) > 100]
print(scaled)
```

## 3. Comparison Operators  (result is always True or False)

```python
a = 10
b = 20
print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= 10)
print(a <= 9)
```

```python
# = is assignment , == is comparison
x = 5          # assignment
print(x == 5)  # comparison
#if x = 5:     # SyntaxError
```

```python
# chained comparison :  a < x < b  means  (a < x) and (x < b)
x = 7
print(1 < x < 10)
print(1 < x < 5)
print(0 <= 0.73 <= 1)        # check that a probability is valid

age = 25
print(18 <= age < 65)
```

```python
# comparing different types
print(3 == 3.0)        # True  (numbers are compared by value)
print(True == 1)       # True  (bool is a subclass of int)
print("3" == 3)        # False (no error , just not equal)
#print("3" < 3)        # TypeError: '<' not supported between instances of 'str' and 'int'
```

```python
# comparing strings , lists and tuples : lexicographic (item by item)
print("apple" < "banana")
print("Zebra" < "apple")         # uppercase letters have smaller unicode codes
print([1, 2, 3] < [1, 3])
print((2026, 5, 1) < (2026, 11, 7))   # dates as tuples
print("10" < "9")                # string comparison , not numeric !
```

```python
# NEVER compare floats with ==
print(0.1 + 0.2 == 0.3)
import math
print(math.isclose(0.1 + 0.2, 0.3))
print(abs((0.1 + 0.2) - 0.3) < 1e-9)    # manual tolerance
```

## 4. Logical Operators  (and  or  not)

```python
# truth table
for p in (True, False):
    for q in (True, False):
        print(f"{p!s:5} {q!s:5} | and={p and q!s:5} or={p or q!s:5}")
print("not True =", not True)
```

```python
# and / or return one of the OPERANDS , not always True/False
print(0 and 5)          # 0     (first falsy value)
print(3 and 5)          # 5     (all truthy ---> last value)
print(0 or 5)           # 5     (first truthy value)
print("" or "default")  # default
print(None or [] or 0)  # 0     (all falsy ---> last value)
```

```python
# Application : default value with  or
user_name = ""
print("Hello,", user_name or "Guest")

# careful : 0 is falsy , so  or  replaces a VALID zero !
dropout = 0.0
print(dropout or 0.5)                            # 0.5  (wrong!)
print(dropout if dropout is not None else 0.5)   # 0.0  (correct)
```

```python
# short-circuit evaluation : the second operand is evaluated ONLY if needed
def check(name, value):
    print("  evaluating", name)
    return value

print(check("A", False) and check("B", True))
print("---")
print(check("A", True) or check("B", True))
```

```python
# Application : short-circuit protects us from errors
data = []
if len(data) > 0 and data[0] > 10:      # data[0] is never evaluated for an empty list
    print("first item is large")
else:
    print("empty or small")

x = 0
print(x != 0 and 10 / x > 1)             # no ZeroDivisionError
```

```python
# precedence :  not  >  and  >  or
print(True or False and False)       # True or (False and False)
print((True or False) and False)
print(not 1 == 2)                    # not (1 == 2)
```

## 5. Identity Operators  (is   is not)

```python
# == compares VALUES , is compares IDENTITY (same object in memory)
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b, a is b)
print(a == c, a is c)
print(id(a), id(b), id(c))
```

```python
# use  is  ONLY for singletons : None , True , False
result = None
print(result is None)
print(result is not None)

x = 1000
y = int("1000")
print(x == y)       # True
print(x is y)       # False here (do NOT rely on 'is' for numbers or strings)
```

## 6. Membership Operators  (in   not in)

```python
print(3 in [1, 2, 3])                  # list  : O(n)
print("py" in "python")                # str   : substring search
print("age" in {"name": 1, "age": 2})  # dict  : checks the KEYS , O(1)
print(2 in {"name": 1, "age": 2}.values())
print("c" not in {"a", "b"})           # set   : O(1)
print(5 in range(0, 10, 5))            # range : computed , O(1)
```

## 7. Bitwise Operators  (& | ^ ~ << >>)  work on the BITS of integers

```python
a = 12      # 1100
b = 10      # 1010
print(f"a      = {a:04b}")
print(f"b      = {b:04b}")
print(f"a & b  = {a & b:04b}  ({a & b})")    # AND
print(f"a | b  = {a | b:04b}  ({a | b})")    # OR
print(f"a ^ b  = {a ^ b:04b}  ({a ^ b})")    # XOR
print(f"~a     = {~a}")                       # NOT : ~x == -x - 1
print(f"a << 1 = {a << 1}")                   # shift left  : x * 2
print(f"a >> 2 = {a >> 2}")                   # shift right : x // 4
```

```python
# Application : bit flags  (many True/False states in ONE integer)
OVERHEAT  = 0b0001
VIBRATION = 0b0010
LOW_OIL   = 0b0100
NO_SIGNAL = 0b1000

status = 0
status |= OVERHEAT          # set a flag
status |= LOW_OIL
print(f"status = {status:04b}")
print("overheat? ", bool(status & OVERHEAT))
print("vibration?", bool(status & VIBRATION))
status &= ~OVERHEAT         # clear a flag
print(f"status = {status:04b}")
```

```python
# & | ^ on sets are union / intersection / symmetric difference (session Set)
s1 = {1, 2, 3}
s2 = {3, 4}
print(s1 & s2, s1 | s2, s1 ^ s2, s1 - s2)
```

## 8. Operators in NumPy and pandas (very important for AI)

```python
import numpy as np
x = np.array([1, 2, 3])
print([1, 2, 3] * 2)      # list : repetition
print(x * 2)              # array : element-wise multiplication
print(x + x)
print(x > 1)              # element-wise comparison ---> boolean array
print(x[x > 1])           # boolean indexing (filtering)
```

```python
# @ matrix multiplication  (the core operation of neural networks)
W = np.array([[0.2, 0.8],
              [0.5, 0.1],
              [0.9, 0.4]])        # weights : 3 neurons x 2 inputs
inp = np.array([1.0, 2.0])        # one input sample
b = np.array([0.1, 0.1, 0.1])     # bias
z = W @ inp + b                   # one dense layer : z = Wx + b
print(z)
print(np.maximum(z - 1, 0))       # ReLU-like activation on (z - 1)
```

```python
# in NumPy/pandas use  & | ~  (NOT and / or / not) and ALWAYS use parentheses
ages = np.array([15, 22, 35, 70])
mask = (ages >= 18) & (ages < 65)
print(mask, ages[mask])
#ages >= 18 and ages < 65      # ValueError: The truth value of an array ... is ambiguous
#ages >= 18 & ages < 65        # wrong precedence : & is evaluated before >= and <
```

## 9. Operator Precedence (high ---> low)

| level | operators |
|---|---|
| 1 | `( )` |
| 2 | `**` |
| 3 | `+x  -x  ~x` (unary) |
| 4 | `*  @  /  //  %` |
| 5 | `+  -` |
| 6 | `<<  >>` |
| 7 | `&` |
| 8 | `^` |
| 9 | `\|` |
| 10 | `==  !=  >  <  >=  <=  is  is not  in  not in` |
| 11 | `not` |
| 12 | `and` |
| 13 | `or` |
| 14 | `if ... else` (conditional expression) |
| 15 | `:=` |

```python
print(2 + 3 * 4)          # 14
print((2 + 3) * 4)        # 20
print(10 - 4 - 3)         # 3   left to right : (10 - 4) - 3
print(2 ** 3 ** 2)        # 512 right to left
print(-3 ** 2)            # -9
print(5 > 3 and 2 > 4 or not False)   # ((5>3) and (2>4)) or (not False)
```
