# Type Conversion & Type Casting in Python
 the type conversion is the process of converting the value of on data type to another data type. 
In general, Type conversion classified into two types :

    - Implicit Type conversion
    - Explicit Type Conversion (type Casting ) 

## Implicit Type Conversion :
Python interpreter automatically converts one data type to another data type.(converts lower data type to higher data type to avoid data loss)

```python
num1 = 15   
num2 = 4.5 
num3= num1 + num2 
print("Value of num3",num3)
print("Data Type of num3",type(num3))
```

## explicit Type conversion 
the user converts one data type to another data type . user converts higher datatype to lower datatype or lower datatype tp higher datatype


 int()   this function convers any data type to integer
 float()   this function convers any data type to a floating point number
 str()   this function convers any data type to a String

```python
a = int(2)
b = int(3.2)
c= int("6")
d= int("11010",2)    # returns base 2 value
print(a)
print(b)
print(c)
print(d)
```

```python
a = float(2)
b = float(3.2)
c= float("6")
d= float("6.2")    
print(a)
print(b)
print(c)
print(d)
```

```python
a = str("abc")
b = str(3.2)
c= int(6)

print(a)
print(b)
print(c)
```
