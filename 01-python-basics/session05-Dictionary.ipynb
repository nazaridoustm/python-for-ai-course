## Dictionary Sequence in Python

Dictionary is the collection of key-value pairs where the value can be any Python object whereas the keys are immutable python object (numbers, sttring or tuple)
dict = {key1: value1 , key2: value2 , .....}

```python
student ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}
print(student)
print(type(student))
```

```python
#accessing the dictionary values 
student ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}
print("name :" , student["name"])
print("age :" , student["age"])
print("major :" , student["major"])
print("branch :" , student["branch"])
```

```python
#updating dictionary vlues 
student ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}
print("name :" , student["name"])
student["name"] = 'Bannie'
print("name :" , student["name"])
print("age :" , student["age"])
student["age"] = "fifty"
print("age :" , student["age"])
student["city"] = "Tehran"    
print(student)
```

```python
#deleting dictionary vlues 
student ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}
print(student)
del student["name"] 
del student["age"] 
print("age :" , student)
```

# Dictionary Functions 
len()   copy()   get()   keys()   items()    values()   update()    pop()   clear()

```python
# len(dictionary)  is uded to find the length of the given dictionary
student ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}
print("length of Dictionary is :" , len(student))
```

```python
# dictionary.copy() returns another copy of given dictionary 
student1 ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}
student2 = student1.copy()
student1["name"] = 'Bannie'
print(student1)
print(student2)
```

```python
import copy        #deep copy
student1 ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}
student2 = copy.deepcopy(student1)
student1["name"] = 'Bannie'
print(student1)
print(student2)
```

```python
student1 = {
    "name": "maryam",
    "scores": [18, 19, 20]
}

student2=student1.copy()

student2["scores"][0] = 10

print(student1)
print(student2)
```

```python
import copy

student1 = {
    "name": "maryam",
    "scores": [18, 19, 20]
}

student2 = copy.deepcopy(student1)

student2["scores"][0] = 10

print(student1)
print(student2)
```

```python
 #dictionary.get()
student1 ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}
print("name is:" , student1.get("name"))
print("major is:" , student1.get("major"))


# print(student1["city"])
print(student1.get("city"))

print(student1.get("city", "Not Found"))
```

```python
 #dictionary.keys() is used to fetch all the keys form the dictionary 
student1 ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}

for i in student1.keys():
    print(i)
```

```python
 #dictionary.keys() is used to fetch all the keys form the dictionary 
student1 ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}

for i in student1:
    print(i)
```

```python
for i in student1.values():
    print(i)
```

```python
 #dictionary.items()  returns a new view of dictionary. this view is collection of key value tuples 
student1 ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}

for i in student1.items():
    print(i)



#for key, value in student1.items():
#   print(key, ":", value)
```

```python
 #dictionary.values() is used to collect all the values from a dictionary
student1 ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}

for i in student1.values():
    print(i)
```

```python
print(list(student1.keys()))  # keys --->list
```

```python
'''
dictionary.update() updates the dictionary with the key and value pairs . it inserts key/value if if it is not present
and it updates key/value if it is already present in the dictionary 
'''
student1 ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}

student1.update({"phone":"0909090909"})
print(student1)
student1.update({"branch":"Qazvin"})
print(student1)

student1.update({"age":50 , "city":"Tehran"})
print(student1)
```

```python
#dictionary.pop() removes an element from the dictionary 
student1 ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}
student1.pop("age")
print(student1)
#student1.pop("sen")
#student1.pop("sen", None)
print(student1)
```

```python
#dictionary.clear() deletes all the items of the dictionary
student1 ={"name" : "maryam" , "age" :42 , "major" :"AI" , "branch" : "Azad1"}
print(student1)
student1.clear()
print(student1)
```
