```python
s = "lotusdata"

print(s[0])         # l

print(s[8])         # a

print(s[-1])        # a

print(s[0:4])       # lotu

print(s[:4])        # lotu

print(s[-9:-5])     # lotu

print(s[4:])        # sdata

print(s[4:9])       # sdata

print(s[-5:])       # sdata

print(s[-1:])       # a
```

```python
# List  is a collection of values or items
l1 = ["c", "java", "python"]  
l2=[]
l3=[123,"python", 3.7]
l4=[1,2,3,4,5,6]

print(type(l1))     # list
print(l1)
print(l2)
print(l3)
print(l4)
```

```python
# Tuple is a sequence of immutable elements or items  
t1 = ("apple", "banana", "cherry")
t2=()
t3=(666, "python", 3.8)
t4 =("c",)
t5=("c")
print(type(t1))     # tuple
print(t1)
print(type(t1)) 
print(t2)
print(type(t2)) 
print(t3)
print(type(t3)) 
print(t4)
print(type(t4)) 
print(t5)
print(type(t5)) 
```

```python
# Dictionary is the collection of key-value paird where the value can be any python objects
# whereas the keys are immutable python object (numbers , string or tuple)
student = {'id': '123', 'name': 'Maryam' , 'age' : 40 , 'major': 'AI'}  
print(type(student))    # dict
print(student)
print(student['id'])
print(student['name'])
print(student['age'])
print(student['major'])
```
