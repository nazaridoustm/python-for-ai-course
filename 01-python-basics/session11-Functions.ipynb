# Functions in Python
a function is a named , reusable block of code that performs ONE specific task.

    def function_name(parameters):
        """docstring"""
        statements
        return value

why functions ?

    - reuse      : write once , use many times (DRY = Don't Repeat Yourself)
    - abstraction: hide the details behind a name  (sigmoid(x) instead of the formula)
    - testing    : a small function is easy to test
    - structure  : a big program = many small functions working together

## 1. Defining and calling a function

```python
def greet():
    print("Hello , welcome to the AI course")

greet()          # calling the function
greet()          # reuse
```

```python
# parameters (in the definition) and arguments (in the call)
def greet(name):                 # name is a PARAMETER
    print(f"Hello {name}")

greet("Maryam")                  # "Maryam" is an ARGUMENT
greet("Ali")
```

```python
# return sends a value back to the caller
def square(x):
    return x * x

result = square(7)
print(result)
print(square(3) + square(4))     # a function call is an expression
```

```python
# a function WITHOUT return (or with an empty return) returns None
def say_hi():
    print("hi")

value = say_hi()
print("returned value:", value)

# print vs return : print shows a value , return GIVES it back
def add_print(a, b):
    print(a + b)

def add_return(a, b):
    return a + b

x = add_print(2, 3)
y = add_return(2, 3)
print("x =", x, " y =", y)
```

```python
# docstring : the documentation of the function (the first string in the body)
def bmi(weight_kg, height_m):
    """Return the Body Mass Index.

    weight_kg : body weight in kilograms
    height_m  : height in meters
    """
    return weight_kg / height_m ** 2

print(round(bmi(70, 1.75), 1))
print(bmi.__doc__)
#help(bmi)        # shows the docstring in a nice format
```

```python
# a function is an OBJECT : it has a type , an id and can be assigned to a name
print(type(square))
print(square.__name__)
f = square            # no parentheses ---> the function itself , not a call
print(f(5))
print(f is square)
```

## 2. Arguments : positional , keyword , default

```python
def describe(name, age, major):
    print(f"{name} is {age} years old and studies {major}")

describe("Maryam", 42, "AI")                      # positional : order matters
describe(major="AI", name="Maryam", age=42)       # keyword : order does not matter
describe("Maryam", major="AI", age=42)            # mixed : positional FIRST
#describe(name="Maryam", 42, "AI")                # SyntaxError: positional argument follows keyword argument
```

```python
# wrong number of arguments
try:
    describe("Maryam", 42)
except TypeError as e:
    print("TypeError:", e)
```

```python
# default values : the parameter becomes optional
def train(model, epochs=10, lr=0.001):
    print(f"training {model} for {epochs} epochs with lr={lr}")

train("resnet")
train("resnet", 50)
train("resnet", lr=0.01)          # skip epochs , change only lr
```

```python
# PITFALL : mutable default value is created ONCE (when the function is defined)
def add_item_wrong(item, bucket=[]):
    bucket.append(item)
    return bucket

print(add_item_wrong(1))
print(add_item_wrong(2))          # [1, 2]  ---> the SAME list is shared !

def add_item(item, bucket=None):  # the correct pattern
    if bucket is None:
        bucket = []
    bucket.append(item)
    return bucket

print(add_item(1))
print(add_item(2))
```

```python
# keyword-only parameters : everything after *  MUST be given by name
def split_data(data, *, test_size=0.2, shuffle=True):
    n_test = int(len(data) * test_size)
    return data[:-n_test], data[-n_test:]

train_part, test_part = split_data(list(range(10)), test_size=0.3)
print(train_part, test_part)
#split_data(list(range(10)), 0.3)     # TypeError: takes 1 positional argument but 2 were given
```

```python
# positional-only parameters : everything before /  MUST be given by position (Python 3.8+)
def power(base, exponent, /):
    return base ** exponent

print(power(2, 10))
#power(base=2, exponent=10)    # TypeError: got some positional-only arguments passed as keyword arguments
```

## 3. *args and **kwargs : any number of arguments

```python
# *args collects extra POSITIONAL arguments into a TUPLE
def total(*args):
    print("args =", args, type(args))
    return sum(args)

print(total(1, 2, 3))
print(total(10, 20))
print(total())
```

```python
# **kwargs collects extra KEYWORD arguments into a DICT
def show_config(**kwargs):
    print(type(kwargs))
    for key, value in kwargs.items():
        print(f"  {key:>10} = {value}")

show_config(lr=0.001, batch_size=32, optimizer="adam")
```

```python
# all together : the order is  normal , *args , keyword-only , **kwargs
def log_metrics(epoch, *values, sep=" | ", **named):
    parts = [f"epoch {epoch}"] + [str(v) for v in values] + [f"{k}={v}" for k, v in named.items()]
    print(sep.join(parts))

log_metrics(3, "train", acc=0.91, loss=0.24)
log_metrics(4, sep=" ; ", f1=0.88)
```

```python
# unpacking in the CALL :  * for lists/tuples , ** for dicts
def describe(name, age, major):
    print(f"{name} | {age} | {major}")

person = ["Sara", 25, "Data Science"]
info = {"name": "Reza", "age": 30, "major": "Robotics"}
describe(*person)
describe(**info)
```

```python
# forwarding pattern : pass everything to another function (used in wrappers / decorators)
def safe_divide(a, b):
    return a / b

def call_and_report(func, *args, **kwargs):
    result = func(*args, **kwargs)
    print(f"{func.__name__}{args} {kwargs} ---> {result}")
    return result

call_and_report(safe_divide, 10, 4)
call_and_report(safe_divide, a=9, b=3)
```

## 4. Return values

```python
# returning several values = returning ONE tuple
def stats(numbers):
    n = len(numbers)
    mean = sum(numbers) / n
    std = (sum((x - mean) ** 2 for x in numbers) / n) ** 0.5
    return min(numbers), max(numbers), mean, std

result = stats([4, 8, 15, 16, 23, 42])
print(result, type(result))
lo, hi, mean, std = stats([4, 8, 15, 16, 23, 42])      # unpacking
print(f"min={lo} max={hi} mean={mean:.2f} std={std:.2f}")
```

```python
# return ends the function immediately (guard clauses)
def safe_mean(numbers):
    if not numbers:
        return None             # early return
    return sum(numbers) / len(numbers)

print(safe_mean([1, 2, 3]))
print(safe_mean([]))
```

```python
# a function should return ONE type of result (here : always a dict) ---> easier to use
def evaluate(y_true, y_pred):
    correct = sum(t == p for t, p in zip(y_true, y_pred, strict=True))
    return {"accuracy": correct / len(y_true), "n": len(y_true), "errors": len(y_true) - correct}

print(evaluate([1, 0, 1, 1], [1, 0, 0, 1]))
```

## 5. Scope : where can a name be seen ?  (LEGB rule)

Python searches a name in this order :

    L : Local       ---> inside the current function
    E : Enclosing   ---> inside the outer function (for nested functions)
    G : Global      ---> at the top level of the module / notebook
    B : Built-in    ---> len , print , sum , ...

```python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print("inner sees:", x)
    inner()
    print("outer sees:", x)

outer()
print("module sees:", x)
```

```python
# a local variable does not exist outside the function
def f():
    secret = 42
    return secret

f()
try:
    print(secret)
except NameError as e:
    print("NameError:", e)
```

```python
# reading a global is allowed , but ASSIGNING to it inside a function creates a NEW local name
counter = 0

def increment_wrong():
    counter += 1            # Python thinks counter is LOCAL (because we assign to it)

try:
    increment_wrong()
except UnboundLocalError as e:
    print("UnboundLocalError:", e)
```

```python
# global : modify the module-level variable (use rarely !)
counter = 0

def increment():
    global counter
    counter += 1

increment(); increment()
print(counter)

# better : pass the value in and return the new value (no hidden state)
def increment_pure(c):
    return c + 1

c = 0
c = increment_pure(c)
c = increment_pure(c)
print(c)
```

```python
# nonlocal : modify a variable of the ENCLOSING function
def make_counter():
    count = 0
    def step():
        nonlocal count
        count += 1
        return count
    return step

next_id = make_counter()
print(next_id(), next_id(), next_id())
```

```python
# shadowing a built-in name breaks it
def bad_sum(values):
    sum = 0                 # local name 'sum' hides the built-in sum()
    for v in values:
        sum += v
    return sum

print(bad_sum([1, 2, 3]))
print(sum([1, 2, 3]))       # still fine OUTSIDE the function
```

## 6. Functions are first-class objects

```python
# 1) pass a function as an argument
def apply_to_all(func, values):
    return [func(v) for v in values]

print(apply_to_all(square, [1, 2, 3]))
print(apply_to_all(str.upper, ["ai", "ml"]))
print(apply_to_all(len, ["python", "c"]))
```

```python
# 2) store functions in a dictionary (registry / dispatch table)
import math

def relu(x):
    return max(0.0, x)

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def tanh(x):
    return math.tanh(x)

ACTIVATIONS = {"relu": relu, "sigmoid": sigmoid, "tanh": tanh}

for name, fn in ACTIVATIONS.items():
    print(f"{name:>8}:", [round(fn(v), 3) for v in (-2.0, 0.0, 2.0)])
```

```python
# 3) return a function from a function (function factory)
def make_threshold_classifier(threshold):
    def classify(prob):
        return 1 if prob >= threshold else 0
    return classify

strict = make_threshold_classifier(0.9)
lenient = make_threshold_classifier(0.3)
probs = [0.2, 0.5, 0.95]
print("strict :", [strict(p) for p in probs])
print("lenient:", [lenient(p) for p in probs])
```

## 7. lambda : small anonymous functions

```python
# lambda parameters: expression     (only ONE expression , no statements)
double = lambda x: x * 2
add = lambda a, b: a + b
print(double(5), add(2, 3))

# the same as
def double_def(x):
    return x * 2
print(double_def(5))
```

```python
# lambda is mostly used as a short argument : key= in sorted / max / min
models = [("logreg", 0.871, 12), ("rf", 0.912, 45), ("xgb", 0.925, 30), ("mlp", 0.925, 8)]

print(sorted(models, key=lambda m: m[1], reverse=True))       # by accuracy
print(max(models, key=lambda m: (m[1], -m[2])))                # best accuracy , then fastest
print(min(models, key=lambda m: m[2]))                         # fastest
```

```python
# map , filter  (return lazy iterators)
nums = [1, 2, 3, 4, 5, 6]
print(list(map(lambda x: x ** 2, nums)))
print(list(filter(lambda x: x % 2 == 0, nums)))

# the same with comprehensions (usually more readable in Python)
print([x ** 2 for x in nums])
print([x for x in nums if x % 2 == 0])

# map with a named function is clean
print(list(map(int, ["10", "20", "30"])))
```

```python
# reduce : combine all items into ONE value
from functools import reduce
print(reduce(lambda acc, x: acc * x, [1, 2, 3, 4, 5]))      # 5! = 120
print(reduce(lambda a, b: a if a > b else b, [3, 9, 2]))     # max
```

## 8. Closures

```python
# a closure is an inner function that REMEMBERS variables of the outer function
def make_scaler(mean, std):
    def scale(x):
        return (x - mean) / std
    return scale

data = [2.0, 4.0, 6.0, 8.0]
m = sum(data) / len(data)
s = (sum((x - m) ** 2 for x in data) / len(data)) ** 0.5
scaler = make_scaler(m, s)                   # "fit" once on training data
print([round(scaler(x), 3) for x in data])
print(round(scaler(10.0), 3))                # "transform" new data with the SAME mean/std
print(scaler.__closure__[0].cell_contents)   # the remembered value
```

```python
# PITFALL : late binding in loops
funcs = [lambda: i for i in range(3)]
print([f() for f in funcs])           # [2, 2, 2]  ---> all see the LAST i

funcs = [lambda i=i: i for i in range(3)]
print([f() for f in funcs])           # [0, 1, 2]  ---> default value captured NOW
```

## 9. Decorators : add behaviour to a function without changing its code

```python
import time
import functools

def timer(func):
    @functools.wraps(func)                    # keep the name and docstring of func
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed = (time.perf_counter() - start) * 1000
        print(f"[timer] {func.__name__} took {elapsed:.2f} ms")
        return result
    return wrapper

@timer                                        # same as: slow_sum = timer(slow_sum)
def slow_sum(n):
    """sum of 0..n-1 with a python loop"""
    total = 0
    for i in range(n):
        total += i
    return total

print(slow_sum(1_000_000))
print(slow_sum.__name__, "|", slow_sum.__doc__)
```

```python
# a decorator WITH arguments : retry a function that may fail (e.g. a network / API call)
def retry(times=3):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            for attempt in range(1, times + 1):
                try:
                    return func(*args, **kwargs)
                except ConnectionError as e:
                    print(f"  attempt {attempt} failed: {e}")
            raise RuntimeError(f"{func.__name__} failed after {times} attempts")
        return wrapper
    return decorator

calls = {"n": 0}

@retry(times=4)
def flaky_api():
    calls["n"] += 1
    if calls["n"] < 3:                        # fails twice , then works
        raise ConnectionError("timeout")
    return "response OK"

print(flaky_api())
```

```python
# built-in decorator : lru_cache (memoization)
from functools import lru_cache

def fib_slow(n):
    return n if n < 2 else fib_slow(n - 1) + fib_slow(n - 2)

@lru_cache(maxsize=None)
def fib_fast(n):
    return n if n < 2 else fib_fast(n - 1) + fib_fast(n - 2)

t = time.perf_counter(); a = fib_slow(27); t_slow = time.perf_counter() - t
t = time.perf_counter(); b = fib_fast(27); t_fast = time.perf_counter() - t
print(a, b)
print(f"without cache: {t_slow * 1000:8.2f} ms")
print(f"with cache   : {t_fast * 1000:8.3f} ms")
print(fib_fast.cache_info())
print(fib_fast(200))                          # instant , impossible without the cache
```

## 10. Recursion : a function that calls itself

```python
# every recursive function needs : 1) a BASE CASE  2) a step that moves towards it
def factorial(n):
    if n <= 1:                 # base case
        return 1
    return n * factorial(n - 1)

print(factorial(5))
print(factorial(20))
```

```python
# without a reachable base case ---> RecursionError
import sys
print("recursion limit:", sys.getrecursionlimit())

def no_base(n):
    return no_base(n + 1)

try:
    no_base(0)
except RecursionError as e:
    print("RecursionError:", e)
```

```python
# recursion is natural for TREE-shaped data : nested JSON / config
config = {
    "model": {"name": "vit", "layers": 12, "head": {"dropout": 0.1, "classes": 10}},
    "train": {"lr": 3e-4, "epochs": 30},
}

def flatten(d, prefix=""):
    flat = {}
    for key, value in d.items():
        full = f"{prefix}.{key}" if prefix else key
        if isinstance(value, dict):
            flat.update(flatten(value, full))      # recursive call
        else:
            flat[full] = value
    return flat

for k, v in flatten(config).items():
    print(f"{k:<22} {v}")
```

```python
# a decision tree stored as nested dicts , predicted with recursion
tree = {
    "feature": "has_contract", "threshold": 0.5,
    "left":  {"feature": "tenure", "threshold": 6,
              "left":  {"feature": "charge", "threshold": 50, "left": {"label": "stay"}, "right": {"label": "churn"}},
              "right": {"feature": "charge", "threshold": 90, "left": {"label": "stay"}, "right": {"label": "churn"}}},
    "right": {"label": "stay"},
}

def predict(node, x):
    if "label" in node:                          # base case : a leaf
        return node["label"]
    branch = "left" if x[node["feature"]] < node["threshold"] else "right"
    return predict(node[branch], x)              # go down one level

customers = [
    {"has_contract": 0, "tenure": 3, "charge": 70},
    {"has_contract": 0, "tenure": 3, "charge": 30},
    {"has_contract": 0, "tenure": 24, "charge": 95},
    {"has_contract": 1, "tenure": 24, "charge": 95},
]
for c in customers:
    print(c, "--->", predict(tree, c))
```

## 11. Type hints and good practices

```python
# type hints document the expected types (Python does NOT check them at runtime)
def normalize(values: list[float], eps: float = 1e-8) -> list[float]:
    """Scale values to the range [0, 1]."""
    lo, hi = min(values), max(values)
    return [(v - lo) / (hi - lo + eps) for v in values]

print([round(v, 3) for v in normalize([10.0, 20.0, 15.0])])
print(normalize.__annotations__)
print(normalize((1, 3, 2)))      # a TUPLE of INTs also works : hints are NOT enforced at runtime (use mypy to check)
```

```python
# pure function vs function with side effects
def add_tax_pure(prices: list[float], rate: float) -> list[float]:
    return [p * (1 + rate) for p in prices]            # returns NEW data , input unchanged

def add_tax_inplace(prices: list[float], rate: float) -> None:
    for i in range(len(prices)):
        prices[i] *= (1 + rate)                        # MODIFIES the caller's list

prices = [100.0, 200.0]
new_prices = add_tax_pure(prices, 0.25)
print(prices, new_prices)
add_tax_inplace(prices, 0.25)
print(prices)
```

## 12. Applications in AI , Machine Learning and Expert Systems

```python
# Application 1 : softmax and cross-entropy as functions (the output layer of a classifier)
def softmax(logits: list[float]) -> list[float]:
    m = max(logits)                                    # subtract max ---> numerical stability
    exps = [math.exp(z - m) for z in logits]
    total = sum(exps)
    return [e / total for e in exps]

def cross_entropy(probs: list[float], true_class: int, eps: float = 1e-12) -> float:
    return -math.log(max(probs[true_class], eps))

logits = [2.0, 1.0, 0.1]
p = softmax(logits)
print("probabilities:", [round(v, 3) for v in p], " sum =", round(sum(p), 6))
print("loss if true class = 0:", round(cross_entropy(p, 0), 4))
print("loss if true class = 2:", round(cross_entropy(p, 2), 4))
print("stable for big logits:", [round(v, 3) for v in softmax([1000, 999, 998])])
```

```python
# Application 2 : numerical gradient + a GENERIC gradient descent (functions as arguments)
def numerical_grad(f, x, h=1e-6):
    return (f(x + h) - f(x - h)) / (2 * h)            # central difference

def minimize(f, x0, lr=0.1, steps=100):
    x = x0
    for _ in range(steps):
        x -= lr * numerical_grad(f, x)
    return x

f1 = lambda x: (x - 3) ** 2 + 1                       # minimum at x = 3
f2 = lambda x: x ** 4 - 3 * x ** 3 + 2                # minimum at x = 2.25
print("min of f1 at x =", round(minimize(f1, x0=0.0), 4))
print("min of f2 at x =", round(minimize(f2, x0=4.0, lr=0.01, steps=2000), 4))
```

```python
# Application 3 : training loop with CALLBACKS (functions called at each epoch)
def early_stopping(patience):
    state = {"best": float("inf"), "wait": 0}
    def callback(epoch, val_loss):
        if val_loss < state["best"]:
            state["best"], state["wait"] = val_loss, 0
        else:
            state["wait"] += 1
        return state["wait"] >= patience             # True ---> stop
    return callback

def printer(epoch, val_loss):
    print(f"  epoch {epoch}: val_loss={val_loss:.2f}")
    return False

def fit(val_losses, callbacks):
    for epoch, loss in enumerate(val_losses, start=1):
        stop = [cb(epoch, loss) for cb in callbacks]  # call EVERY callback
        if any(stop):
            print(f"  stopped at epoch {epoch}")
            break

fit([0.9, 0.7, 0.65, 0.66, 0.67, 0.64], callbacks=[printer, early_stopping(patience=2)])
```

```python
# Application 4 : expert system with rules as FUNCTIONS (predicates)
RULES = [
    ("possible bearing failure", lambda f: f["temp"] > 85 and f["vibration"] > 7.1),
    ("check cooling system",     lambda f: f["temp"] > 85),
    ("check shaft alignment",    lambda f: f["vibration"] > 7.1),
    ("check for leaks",          lambda f: f["pressure"] < 2.0),
]

def diagnose(facts: dict, rules=RULES) -> list[str]:
    return [name for name, condition in rules if condition(facts)]

print(diagnose({"temp": 92, "vibration": 8.3, "pressure": 3.1}))
print(diagnose({"temp": 70, "vibration": 3.0, "pressure": 1.5}))
print(diagnose({"temp": 65, "vibration": 2.5, "pressure": 3.2}) or ["NORMAL"])
```

```python
# Application 5 : a preprocessing PIPELINE built by composing small functions
def compose(*funcs):
    def pipeline(x):
        for f in funcs:
            x = f(x)
        return x
    return pipeline

PERSIAN_DIGITS = str.maketrans("۰۱۲۳۴۵۶۷۸۹", "0123456789")
strip       = str.strip
lower       = str.lower
fa_digits   = lambda s: s.translate(PERSIAN_DIGITS)
fix_yeh     = lambda s: s.replace("ي", "ی").replace("ك", "ک")
squeeze     = lambda s: " ".join(s.split())

clean = compose(strip, fa_digits, fix_yeh, lower, squeeze)
print(repr(clean("   كيفيت   محصول ۱۰ از ۱۰   GREAT  ")))
```
