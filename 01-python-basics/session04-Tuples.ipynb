**Tuple sequence in python is used to store the sequence of immutable python object
tuple is similar to lists since the value of the items stored in the list can be changed
whereas the tuple is immutable and the value of the items stored in the tuple can not be changed.
a tuple can be written as the collection of camma-separated with the small brackets.**

```python
#tuple sequence in python tuple1 =(value1, value2 , ...)
t1 =()
t2 =(10,20,30.40,60)
t3 =("c", "java", "python")
t4 = (501, "abc" ,19.5)
t5=(90,)
print(t1)
print(t2)
print(t3)
print(t4)
print(t5)
print(type(t5))
```

```python
#tuple operation + * []  [start:end]  in  not in
num = (1,2,3,4,5)
lang = ("python","c","java","php")
print(num + lang)
print(num * 2)
print(lang[2])
print(lang[1:4]) 
print("cpp" in lang)
print(6 not in num)
```

### tuple Functions (len()  max()  min()  tuple() sum()  sorted ()  index() count()  )

```python
# len(tuple ) is used to find the length of tuple 
t1 =()
t2 =(10,20,30.40,60)
t3 =("c", "java", "python")
t4 = (501, "abc" ,19.5)
print(len(t1))
print(len(t2))
print(len(t3))
print(len(t4))
```

```python
t= (1,2,3,(4,5,6))     
print(len(t))
```

```python
# max(tuple ) is used to find the Maximum value of tuple 
t1 =(10,20,30.40,60)
t2 =("c", "javaaaa", "python")
print(" Max of tuple :" , max(t1))
print(" Max of tuple :" , max(t2))
print("the maximum value of this list is :" ,max(t2, key=len))
#print("the maximum value of this list is :" ,max(t1, key=len))
```

```python
# min(tuple ) is used to find the Minimum value of tuple 
t1 =(10,20,30.40,60)
t2 =("c", "java", "python")
print(" Min of tuple :" , min(t1))
print(" Min of tuple :" , min(t2))
print("the maximum value of this list is :" ,min(t2, key=len))
#print("the maximum value of this list is :" ,min(t1, key=len))
```

```python
# sum(tuple ) is used to return sum of the values of all the tuple 
num =(10,20,30, 40,60)
#num =(10,20,30,[ 40,60])
print(" sum of tuple items      :" , sum(num))
print(" sum of tuple items + 100:",sum(num,100 ))   # sum( iterable, start=0)


average = sum(num)/len(num)
print(" Average : ",average)
```

```python
# sorted(tuple ) is used to sort all items of tuple in ascendin order for decending reverse = True 
t1 =(10,20,30,40,60)
t2 =("c", "java", "python")
print( sorted(t1))
print( sorted(t2))
print( type(sorted(t1)))  # sortedreturns the LIST 
print(type(sorted(t2)))
print( sorted(t1, reverse = True))
print( sorted(t2, reverse = True))
```

```python
#tuple.count()   the number of times element appears in the tuple.if the element is not present in the tuple it returns 0.
num = (1,2,2,6,3,4,5,3,7,6,8,9,8,2,2)  
cnt = num.count (2) 
print ("count of 2 is :" ,cnt)
cnt = num.count (3)
print ("count of 3 is :" ,cnt)
cnt = num.count (10)
print ("count of 0 is :" ,cnt)

print (True +True)

t=(1,True,1.0)   
print(t.count(1))    # 1==True==1.0
```

```python
#tuple.index() it takes an argument and returns index of it. if the element is not present, it raises a ValueError

lang =('p','y','t','h','o','n','p','r','o','g','r','a','m') 

print("index of 't'  is:",lang.index('t'))     #first 't' in the tuple
print("index of 'p'  is:",lang.index('p')) 
#print("index of 'z'  is:",lang.index('z'))    # we dont have 'z' so there is a ValuError 
print("index of 'p'  is:",lang.index('p',3,10))# between 3 :9  index(value, start ,end) 
```
