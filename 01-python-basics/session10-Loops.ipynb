# Loops in Python
loops repeat a block of code.

    1. for loop     ---> repeat for EACH item of an iterable (list, str, tuple, set, dict, range, file, ...)
    2. while loop   ---> repeat WHILE a condition is True
    3. loop control ---> break , continue , pass , else
    4. comprehensions and generator expressions

## 1. for loop

```python
# for item in iterable:
#     statements
langs = ["python", "c", "java"]
for lang in langs:
    print(lang)
```

```python
# for works on ANY iterable
for ch in "AI":                       # string ---> characters
    print(ch, end=" ")
print()
for x in (10, 20):                    # tuple
    print(x, end=" ")
print()
for x in sorted({"b", "a", "c"}):     # set (sorted to get a fixed order)
    print(x, end=" ")
print()
```

```python
# range(start, stop, step)  ---> stop is NOT included
print(list(range(5)))
print(list(range(2, 8)))
print(list(range(0, 20, 5)))
print(list(range(10, 0, -3)))         # counting down
print(list(range(5, 2)))              # empty : start > stop with positive step
```

```python
# sum of 1..100  (Gauss)
total = 0
for i in range(1, 101):
    total += i
print(total, "=", 100 * 101 // 2)
```

```python
# loop over a dictionary
student = {"name": "Maryam", "age": 42, "major": "AI"}
for key in student:                   # keys
    print(key, end=" ")
print()
for value in student.values():
    print(value, end=" ")
print()
for key, value in student.items():    # key-value pairs (most common)
    print(f"{key:>6}: {value}")
```

## 2. enumerate , zip , reversed

```python
# enumerate : index + value  (instead of range(len(...)))
models = ["logreg", "rf", "xgb"]
for i, m in enumerate(models):
    print(i, m)
for rank, m in enumerate(models, start=1):
    print(f"#{rank} {m}")
```

```python
# zip : loop over several iterables in parallel
names = ["Ali", "Sara", "Reza"]
scores = [17, 19, 15]
for name, score in zip(names, scores):
    print(name, score)

# zip stops at the SHORTEST iterable (silently!)
print(list(zip([1, 2, 3], ["a", "b"])))
# strict=True (Python 3.10+) raises an error if lengths differ
#list(zip([1, 2, 3], ["a", "b"], strict=True))    # ValueError
```

```python
# zip + dict : build a mapping
label2id = dict(zip(["cat", "dog", "bird"], range(3)))
print(label2id)

# unzip with zip(*...)
pairs = [("a", 1), ("b", 2), ("c", 3)]
letters, numbers = zip(*pairs)
print(letters, numbers)
```

```python
# reversed() and sorted() do NOT change the original list
nums = [3, 1, 2]
for n in reversed(nums):
    print(n, end=" ")
print()
for n in sorted(nums, reverse=True):
    print(n, end=" ")
print()
print(nums)
```

## 3. while loop

```python
# while condition:
#     statements      (the condition MUST become False at some point)
count = 1
while count <= 5:
    print("count =", count)
    count += 1                         # without this ---> infinite loop
```

```python
# while is used when the number of repetitions is NOT known in advance
# example : how many years until an investment doubles at 18% per year ?
money, years = 100.0, 0
while money < 200:
    money *= 1.18
    years += 1
print(f"{years} years ---> {money:.1f}")
```

```python
# Collatz sequence : nobody knows in advance how long it is
n = 27
steps = 0
while n != 1:
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    steps += 1
print("steps:", steps)
```

```python
# input validation pattern (simulated input list instead of input())
fake_inputs = iter(["abc", "", "3.5", "42"])
while True:
    raw = next(fake_inputs)            # in a real program : raw = input("Enter an integer: ")
    try:
        value = int(raw)
        break
    except ValueError:
        print(f"{raw!r} is not an integer, try again")
print("accepted:", value)
```

## 4. break , continue , pass , else

```python
# break : exit the loop immediately
for n in [4, 8, 15, 16, 23, 42]:
    if n > 20:
        print("found first number > 20 :", n)
        break
    print("checked", n)
```

```python
# continue : skip the rest of THIS iteration and go to the next one
readings = [72, None, 75, -999, 80]      # None and -999 are invalid sensor values
valid = []
for r in readings:
    if r is None or r < 0:
        continue
    valid.append(r)
print(valid)
```

```python
# pass : does nothing (placeholder for code you will write later)
for i in range(3):
    pass
print("loop finished")
```

```python
# else on a loop : runs ONLY if the loop finished WITHOUT break
def find(items, target):
    for i, x in enumerate(items):
        if x == target:
            print(f"found {target} at index {i}")
            break
    else:
        print(f"{target} not found")

find([5, 3, 9], 3)
find([5, 3, 9], 7)
```

```python
# prime check with for-else
for n in range(2, 20):
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            break
    else:
        print(n, end=" ")
print()
```

## 5. Nested loops

```python
# multiplication table
for i in range(1, 4):
    for j in range(1, 6):
        print(f"{i * j:3}", end="")
    print()
```

```python
# loop over a 2D list (matrix)
matrix = [[1, 2, 3],
          [4, 5, 6]]
for row in matrix:
    for value in row:
        print(value, end=" ")
    print()

# transpose with nested loops
rows, cols = len(matrix), len(matrix[0])
transposed = [[matrix[r][c] for r in range(rows)] for c in range(cols)]
print(transposed)
```

```python
# break exits ONLY the inner loop
for i in range(3):
    for j in range(3):
        if j == 1:
            break
        print(i, j)
```

```python
# exit BOTH loops : use a function with return (cleanest)
def find_pair(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return i, j
    return None

print(find_pair([2, 7, 11, 15], 18))
print(find_pair([2, 7, 11, 15], 100))
```

## 6. Common mistakes

```python
# mistake 1 : modifying a list while looping over it  (skips items silently)
x = [1, 2, 2, 3]
for v in x:
    if v == 2:
        x.remove(v)
print("wrong :", x)

x = [1, 2, 2, 3]
x = [v for v in x if v != 2]
print("right :", x)
```

```python
# mistake 2 : changing the loop variable does NOT change the list
nums = [1, 2, 3]
for n in nums:
    n = n * 10
print(nums)

for i in range(len(nums)):        # change by index
    nums[i] *= 10
print(nums)
```

```python
# mistake 3 : the loop variable still exists AFTER the loop
for k in range(3):
    pass
print("k after loop =", k)

# mistake 4 : range(len(...)) when you only need the values
names = ["a", "b"]
for i in range(len(names)):       # works but not pythonic
    print(names[i], end=" ")
for name in names:                # better
    print(name, end=" ")
print()
```

## 7. Comprehensions and generator expressions

```python
# list comprehension : [expression for item in iterable if condition]
squares = [n ** 2 for n in range(10)]
evens = [n for n in range(10) if n % 2 == 0]
labels = ["pos" if s > 0.5 else "neg" for s in [0.9, 0.2, 0.6]]
print(squares)
print(evens)
print(labels)
```

```python
# the same as a normal loop
evens2 = []
for n in range(10):
    if n % 2 == 0:
        evens2.append(n)
print(evens2 == evens)
```

```python
# set and dict comprehensions
words = ["Python", "python", "AI", "ai", "ML"]
print({w.lower() for w in words})
print({w: len(w) for w in words})
print({k: v for k, v in {"acc": 0.94, "f1": 0.88, "auc": 0.97}.items() if v > 0.9})
```

```python
# nested comprehension : flatten a matrix
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [x for row in matrix for x in row]       # same order as two nested for loops
print(flat)
```

```python
# generator expression ( ) : produces values one by one (lazy) ---> almost no memory
import sys
big_list = [n * n for n in range(1_000_000)]
big_gen = (n * n for n in range(1_000_000))
print("list :", sys.getsizeof(big_list), "bytes")
print("gen  :", sys.getsizeof(big_gen), "bytes")
print(sum(n * n for n in range(1_000_000)))    # no extra brackets needed inside a function call
```

## 8. How for works : the iterator protocol

```python
# for calls iter() once , then next() until StopIteration
langs = ["python", "c"]
it = iter(langs)
print(next(it))
print(next(it))
try:
    next(it)
except StopIteration:
    print("StopIteration ---> the for loop stops here")
```

```python
# an iterator can be consumed only ONCE
gen = (x for x in range(3))
print(list(gen))
print(list(gen))       # empty !
```

## 9. Performance : loop vs built-in vs NumPy (vectorization)

```python
import time
import numpy as np
N = 1_000_000
data = list(range(N))
arr = np.arange(N)

t = time.perf_counter()
total = 0
for v in data:
    total += v
t_loop = time.perf_counter() - t

t = time.perf_counter(); total2 = sum(data); t_sum = time.perf_counter() - t
t = time.perf_counter(); total3 = int(arr.sum()); t_np = time.perf_counter() - t

print(total == total2 == total3)
print(f"python for loop : {t_loop * 1000:7.2f} ms")
print(f"built-in sum()  : {t_sum * 1000:7.2f} ms")
print(f"numpy .sum()    : {t_np * 1000:7.2f} ms")
```

## 10. Applications in AI , Machine Learning and Expert Systems

```python
# Application 1 : Gradient Descent for linear regression  y = w*x + b  (from scratch)
X = [1, 2, 3, 4, 5]
Y = [3.1, 4.9, 7.2, 8.8, 11.1]          # roughly y = 2x + 1
w, b = 0.0, 0.0
lr = 0.02
n = len(X)

for epoch in range(1, 2001):
    dw = db = 0.0
    loss = 0.0
    for x, y in zip(X, Y):
        error = (w * x + b) - y
        loss += error ** 2
        dw += 2 * error * x
        db += 2 * error
    w -= lr * dw / n
    b -= lr * db / n
    if epoch in (1, 10, 100, 500, 2000):
        print(f"epoch {epoch:4d}  loss={loss / n:.4f}  w={w:.3f}  b={b:.3f}")
```

```python
# Application 2 : Forward Chaining inference engine (expert system)
# keep firing rules until NO new fact can be derived
rules = [
    ({"is_penguin"}, "lives_in_cold_climate"),
    ({"is_bird", "cannot_fly", "swims"}, "is_penguin"),
    ({"is_bird", "can_talk"}, "is_parrot"),
    ({"has_feathers"}, "is_bird"),
]
facts = {"has_feathers", "cannot_fly", "swims"}

changed = True
iteration = 0
while changed:
    changed = False
    iteration += 1
    for conditions, conclusion in rules:
        if conditions <= facts and conclusion not in facts:
            facts.add(conclusion)
            print(f"iteration {iteration}: {sorted(conditions)} ---> {conclusion}")
            changed = True
print("final facts:", sorted(facts))
```

```python
# Application 3 : Breadth-First Search (shortest path in a graph , e.g. a warehouse robot)
from collections import deque
graph = {
    "A": ["B", "C"], "B": ["D"], "C": ["D", "E"],
    "D": ["F"], "E": ["F"], "F": [],
}
def bfs(start, goal):
    queue = deque([[start]])
    visited = {start}
    while queue:
        path = queue.popleft()
        node = path[-1]
        if node == goal:
            return path
        for nxt in graph[node]:
            if nxt not in visited:
                visited.add(nxt)
                queue.append(path + [nxt])
    return None

print(bfs("A", "F"))
print(bfs("E", "A"))
```

```python
# Application 4 : Monte Carlo estimation of pi  (random simulation)
import random
random.seed(42)                       # reproducible results
inside = 0
N = 200_000
for _ in range(N):                    # _ means : the loop variable is not needed
    x, y = random.random(), random.random()
    if x * x + y * y <= 1:
        inside += 1
print("pi ~", 4 * inside / N)
```

```python
# Application 5 : moving average of sensor data with a sliding window
from collections import deque
stream = [20, 21, 22, 40, 23, 22, 21, 50, 22]
window = deque(maxlen=3)                 # the last 3 values
for t, value in enumerate(stream):
    if window:                           # compare with the PREVIOUS values
        avg = sum(window) / len(window)
        flag = "  <--- anomaly" if value > 1.5 * avg else ""
        print(f"t={t}  value={value:3}  avg_of_last_3={avg:5.1f}{flag}")
    window.append(value)                 # the oldest value is dropped automatically
```
