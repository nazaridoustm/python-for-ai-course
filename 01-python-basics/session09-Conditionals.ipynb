# Conditional Statements in Python
conditional statements let the program take DECISIONS : a block of code runs only if a condition is True.

    1. if
    2. if ... else
    3. if ... elif ... else
    4. nested if
    5. conditional expression (ternary operator)
    6. match ... case   (Python 3.10+)

Python uses INDENTATION (4 spaces) , not { } , to define the block of each condition.

## 1. if statement

```python
# if condition:
#     statements      <--- indented block
temperature = 92
if temperature > 85:
    print("Warning : temperature is high")
    print("send alert to the operator")
print("this line always runs (it is outside the if block)")
```

```python
# the condition can be ANY expression : Python uses its truthiness (bool(value))
items = []
if items:
    print("list has items")
if not items:
    print("list is empty")

name = "Maryam"
if name:
    print("name is not empty")
```

## 2. if ... else

```python
score = 0.73
threshold = 0.5
if score >= threshold:
    label = "spam"
else:
    label = "not spam"
print(f"score = {score}  --->  {label}")
```

```python
# even or odd
n = 17
if n % 2 == 0:
    print(n, "is even")
else:
    print(n, "is odd")
```

## 3. if ... elif ... else

```python
# only the FIRST True branch runs , the rest are skipped
grade = 17.5
if grade >= 18:
    result = "A"
elif grade >= 15:
    result = "B"
elif grade >= 12:
    result = "C"
elif grade >= 10:
    result = "D"
else:
    result = "F"
print(grade, "--->", result)
```

```python
# the ORDER of elif matters !  (wrong order)
grade = 19
if grade >= 10:
    result = "D"          # 19 >= 10 is True ---> stops here
elif grade >= 18:
    result = "A"          # never reached
print(grade, "--->", result, "   (wrong!)")
```

```python
# if + if  is NOT the same as  if + elif
x = 15
print("if / if :")
if x > 10:
    print("  greater than 10")
if x > 5:
    print("  greater than 5")       # also runs

print("if / elif :")
if x > 10:
    print("  greater than 10")
elif x > 5:
    print("  greater than 5")       # skipped
```

```python
# Application : BMI category (health app)
weight, height = 70, 1.75
bmi = weight / height ** 2
if bmi < 18.5:
    category = "underweight"
elif bmi < 25:
    category = "normal"
elif bmi < 30:
    category = "overweight"
else:
    category = "obese"
print(f"BMI = {bmi:.1f} ---> {category}")
```

## 4. Nested if  and  logical operators

```python
# nested if
age = 30
income = 45_000
if age >= 18:
    if income >= 30_000:
        print("loan application : approved for review")
    else:
        print("loan application : income too low")
else:
    print("loan application : applicant is a minor")
```

```python
# the same logic with  and  : flatter and easier to read
age, income = 30, 45_000
if age >= 18 and income >= 30_000:
    print("approved for review")
elif age >= 18:
    print("income too low")
else:
    print("applicant is a minor")
```

```python
# guard clause : check the bad cases FIRST and exit early (less nesting)
def approve(age, income, has_debt):
    if age < 18:
        return "rejected : minor"
    if income < 30_000:
        return "rejected : low income"
    if has_debt:
        return "manual review : existing debt"
    return "approved"

print(approve(16, 50_000, False))
print(approve(30, 20_000, False))
print(approve(30, 50_000, True))
print(approve(30, 50_000, False))
```

## 5. Common mistakes

```python
# mistake 1 :  x == 1 or 2   is ALWAYS True
x = 5
print(x == 1 or 2)        # (x == 1) or 2  --->  False or 2  --->  2 (truthy!)
if x == 1 or 2:
    print("  wrong : this always runs")

# correct
print(x == 1 or x == 2)
print(x in (1, 2))        # best
```

```python
# mistake 2 : comparing floats with ==
total = 0.1 + 0.2
if total == 0.3:
    print("equal")
else:
    print("not equal !", total)

import math
if math.isclose(total, 0.3):
    print("isclose ---> equal")
```

```python
# mistake 3 : if value:  when 0 is a VALID value
learning_rate = 0.0          # user really wants 0
if learning_rate:
    print("use", learning_rate)
else:
    print("no learning rate given ?!  (wrong : 0.0 is falsy)")

if learning_rate is not None:
    print("use", learning_rate, "(correct)")
```

```python
# mistake 4 : None must be checked with  is
value = None
print(value is None)          # correct
print(value == None)          # works but not recommended (PEP 8)
```

## 6. Conditional Expression (ternary operator)

```python
# value_if_true  if condition  else  value_if_false
prob = 0.81
label = "positive" if prob >= 0.5 else "negative"
print(label)

n = 7
print("even" if n % 2 == 0 else "odd")

# inside a list comprehension
scores = [0.2, 0.7, 0.55, 0.9, 0.1]
preds = [1 if s >= 0.5 else 0 for s in scores]
print(preds)
```

```python
# nested ternary : possible but hard to read ---> prefer if/elif
t = 15
state = "cold" if t < 10 else "warm" if t < 25 else "hot"
print(state)
```

```python
# min / max / clamp often replace if/else
x = 1.3
clamped = max(0.0, min(1.0, x))       # keep x in [0, 1]
print(clamped)
relu = max(0, -2.5)                   # ReLU activation
print(relu)
```

## 7. any() and all()  : conditions on many values

```python
readings = [72, 75, 91, 70]
print(any(r > 90 for r in readings))     # at least one is True
print(all(r > 60 for r in readings))     # every one is True

required = ["age", "income", "city"]
row = {"age": 30, "income": 45000}
if not all(col in row for col in required):
    missing = [c for c in required if c not in row]
    print("missing columns:", missing)

print(any([]), all([]))                  # empty : any ---> False , all ---> True
```

## 8. match ... case  (Structural Pattern Matching , Python 3.10+)

```python
# 1) literal patterns  ( | means OR ,  _ is the wildcard = default)
def http_status(code):
    match code:
        case 200:
            return "OK"
        case 400 | 422:
            return "bad request"
        case 404:
            return "not found"
        case 500 | 502 | 503:
            return "server error"
        case _:
            return "unknown"

for c in (200, 422, 404, 503, 999):
    print(c, "--->", http_status(c))
```

```python
# 2) sequence patterns + capture variables + guard (if)
def describe(point):
    match point:
        case (0, 0):
            return "origin"
        case (0, y):
            return f"on the y axis at {y}"
        case (x, 0):
            return f"on the x axis at {x}"
        case (x, y) if x == y:
            return f"on the diagonal at {x}"
        case (x, y):
            return f"point ({x}, {y})"
        case _:
            return "not a 2D point"

for p in [(0, 0), (0, 5), (3, 0), (4, 4), (2, 7), (1, 2, 3)]:
    print(p, "--->", describe(p))
```

```python
# 3) mapping patterns : very useful for JSON / API messages (e.g. LLM tool calls)
def handle(msg):
    match msg:
        case {"role": "user", "content": text}:
            return f"user said: {text}"
        case {"role": "assistant", "tool_call": {"name": name, "args": args}}:
            return f"call tool {name} with {args}"
        case {"role": "assistant", "content": text}:
            return f"assistant said: {text}"
        case _:
            return "unknown message"

print(handle({"role": "user", "content": "hello"}))
print(handle({"role": "assistant", "tool_call": {"name": "weather", "args": {"city": "Tehran"}}}))
print(handle({"role": "assistant", "content": "hi!"}))
print(handle({"role": "system"}))
```

```python
# 4) command parser with  *rest
def run(command):
    match command.split():
        case ["train", model]:
            return f"training {model} with default epochs"
        case ["train", model, epochs] if epochs.isdecimal():
            return f"training {model} for {epochs} epochs"
        case ["predict", *files] if files:
            return f"predicting {len(files)} file(s): {files}"
        case ["quit" | "exit"]:
            return "bye"
        case _:
            return f"unknown command: {command!r}"

for cmd in ["train resnet", "train vit 30", "predict a.jpg b.jpg", "exit", "train vit many"]:
    print(cmd, "--->", run(cmd))
```

## 9. Replacing long if/elif chains with a dictionary

```python
# long chain
def unit_price_if(product):
    if product == "cpu":
        return 300
    elif product == "gpu":
        return 1200
    elif product == "ram":
        return 80
    else:
        return None

# dictionary lookup : shorter , faster for many cases , easy to extend
PRICES = {"cpu": 300, "gpu": 1200, "ram": 80}
def unit_price(product):
    return PRICES.get(product)

print(unit_price_if("gpu"), unit_price("gpu"), unit_price("ssd"))
```

## 10. Applications in AI and Expert Systems

```python
# Application 1 : decision threshold + confusion matrix categories
y_true = [1, 0, 1, 1, 0, 0, 1, 0]
y_prob = [0.9, 0.4, 0.35, 0.8, 0.6, 0.1, 0.55, 0.2]
threshold = 0.5

tp = fp = tn = fn = 0
for t, p in zip(y_true, y_prob):
    pred = 1 if p >= threshold else 0
    if pred == 1 and t == 1:
        tp += 1
    elif pred == 1 and t == 0:
        fp += 1
    elif pred == 0 and t == 0:
        tn += 1
    else:
        fn += 1

precision = tp / (tp + fp) if (tp + fp) else 0.0     # avoid division by zero
recall = tp / (tp + fn) if (tp + fn) else 0.0
print(f"TP={tp} FP={fp} TN={tn} FN={fn}")
print(f"precision={precision:.2f}  recall={recall:.2f}")
```

```python
# Application 2 : a small rule-based expert system (equipment diagnosis)
def diagnose(temp, vibration, pressure):
    if temp > 85 and vibration > 7.1:
        return "CRITICAL : possible bearing failure ---> stop the machine"
    elif temp > 85:
        return "WARNING : check the cooling system"
    elif vibration > 7.1:
        return "WARNING : check shaft alignment"
    elif pressure < 2.0:
        return "WARNING : low pressure ---> check for leaks"
    else:
        return "NORMAL"

samples = [(92, 8.3, 3.1), (90, 4.0, 3.0), (70, 9.0, 3.0), (70, 3.0, 1.5), (65, 2.5, 3.2)]
for s in samples:
    print(s, "--->", diagnose(*s))
```

```python
# Application 3 : simple decision tree written by hand (this is what a trained tree learns)
def predict_churn(tenure_months, monthly_charge, has_contract):
    if has_contract:
        return "stay"
    if tenure_months < 6:
        return "churn" if monthly_charge > 50 else "stay"
    return "churn" if monthly_charge > 90 else "stay"

customers = [(3, 70, False), (3, 30, False), (24, 95, False), (24, 95, True)]
for c in customers:
    print(c, "--->", predict_churn(*c))
```

```python
# Application 4 : early stopping check inside training
best_loss, wait, patience = float("inf"), 0, 2
for epoch, val_loss in enumerate([0.90, 0.71, 0.65, 0.66, 0.67, 0.64], start=1):
    if val_loss < best_loss:
        best_loss, wait = val_loss, 0
        status = "improved"
    else:
        wait += 1
        status = f"no improvement ({wait}/{patience})"
    print(f"epoch {epoch}: val_loss={val_loss:.2f}  {status}")
    if wait >= patience:
        print("early stopping!")
        break
```
