## Frozenset in Python

**frozenset is the immutable version of set: an unordered collection of unique and hashable items.
because it is immutable it is also hashable, so it can be used as a dictionary key or as an element of another set.
there is no literal syntax for frozenset ( {} makes a set or a dict ) and it is created only by frozenset(iterable)**

set  ---> frozenset   is like   list ---> tuple

```python
#frozenset sequence in python   fs = frozenset(iterable)
fs1 = frozenset({"bread", "butter", "milk"})
fs2 = frozenset([1, 2, 3, 2, 1])        # duplicates are removed
fs3 = frozenset()                       # empty frozenset
print(fs1)
print(fs2)
print(fs3)
print(type(fs1))
print(type({}))                         # {} is a dict  not a set and not a frozenset
print(type(set()))
```

```python
# frozenset(sequence) takes any iterable and converts it to frozenset
fs1 = frozenset("Maryam")               # string  -> unique characters (case sensitive)
fs2 = frozenset(("c", "java", "python"))  # tuple
fs3 = frozenset(range(5))               # range
fs4 = frozenset({"name": "maryam", "age": 42})   # dict -> only the KEYS
print(fs1)
print(fs2)
print(fs3)
print(fs4)
```

```python
# frozenset is unordered : no indexing and no slicing
fs = frozenset({"c", "java", "python"})
#print(fs[0])        # TypeError: 'frozenset' object is not subscriptable
for item in fs :     # the order of looping is NOT guaranteed
    print(item)
print(sorted(fs))    # sorted() returns a LIST with a fixed order
```

### frozenset is immutable

```python
# add / remove / discard / pop / clear / update  do NOT exist for frozenset
fs = frozenset({1, 2, 3})
#fs.add(4)          # AttributeError: 'frozenset' object has no attribute 'add'
#fs.remove(1)       # AttributeError: 'frozenset' object has no attribute 'remove'
#fs.clear()         # AttributeError: 'frozenset' object has no attribute 'clear'
print(hasattr(set(), "add"))
print(hasattr(fs, "add"))
```

```python
# |=  does not change the frozenset , it creates a NEW object and moves the name to it
fs = frozenset({1, 2, 3})
old = fs             # keep a reference to the old object
print(id(fs))
fs |= {4}
print(fs)
print(id(fs))        # different id ---> new object (like += on tuple)
print(old)           # the old frozenset did not change

s = {1, 2, 3}        # compare with set
print(id(s))
s |= {4}
print(id(s))         # same id ---> changed in place
```

```python
# frozenset.copy() returns the SAME object (there is no reason to copy an immutable object)
fs = frozenset({1, 2, 3})
fs_copy = fs.copy()
print(fs_copy is fs)

s = {1, 2, 3}
s_copy = s.copy()
print(s_copy is s)
```

### frozenset is hashable

```python
# hash()  set is unhashable but frozenset is hashable
fs = frozenset({1, 2, 3})
print(hash(fs))
#print(hash({1, 2, 3}))     # TypeError: unhashable type: 'set'

# order of items does not change the hash
print(hash(frozenset({1, 2})) == hash(frozenset({2, 1})))

# but the order changes the hash of tuple
print(hash((1, 2)) == hash((2, 1)))
```

```python
# frozenset as a dictionary key
#d = {{1, 2}: "x"}          # TypeError: unhashable type: 'set'
d = {frozenset({1, 2}): "x"}
print(d)
print(d[frozenset({2, 1})])  # same key , different order
```

```python
# set of sets  is only possible with frozenset
#s = {{1}, {2}}             # TypeError: unhashable type: 'set'
s = {frozenset({1}), frozenset({2}), frozenset({2})}
print(s)
print(len(s))
```

### frozenset operators and methods ( | & - ^ )

```python
#union operator
days1 = frozenset({"Monday", "Tuesday", "Wednesday", "Saturday"})
days2 = frozenset({"Tuesday", "Friday", "Saturday", "Sunday"})
print(days1 | days2)
print(days1.union(days2))
print(days1.union(["Thursday"]))     # method accepts any iterable
```

```python
#intersection operator
days1 = frozenset({"Monday", "Tuesday", "Wednesday", "Saturday"})
days2 = frozenset({"Tuesday", "Friday", "Saturday", "Sunday"})
print(days1 & days2)
print(days1.intersection(days2))
```

```python
#difference operator  (A - B)
days1 = frozenset({"Monday", "Tuesday", "Wednesday", "Saturday"})
days2 = frozenset({"Tuesday", "Friday", "Saturday", "Sunday"})
print(days1 - days2)
print(days1.difference(days2))
print(days2 - days1)                 # difference is NOT commutative
```

```python
#symmetric difference operator  (A ^ B)  items that are in exactly one of them
days1 = frozenset({"Monday", "Tuesday", "Wednesday", "Saturday"})
days2 = frozenset({"Tuesday", "Friday", "Saturday", "Sunday"})
print(days1 ^ days2)
print(days1.symmetric_difference(days2))
```

```python
# the type of result comes from the LEFT operand
fs = frozenset({1, 2})
s = {2, 3}
print(type(fs | s))
print(type(s | fs))
print(type(fs & s))
```

```python
# equality between set and frozenset is based on the items
print({1, 2} == frozenset({1, 2}))
print(frozenset({1, 2}) == frozenset({2, 1}))
```

```python
# issubset()  issuperset()  isdisjoint()
fs1 = frozenset({1, 2, 3, 4, 5, 6, 7, 8, 9, 0})
fs2 = frozenset({1, 2, 6, 7, 8, 9})
fs3 = frozenset({100, 200})
print(fs2.issubset(fs1))       # fs2 <= fs1
print(fs2 <= fs1)
print(fs2 < fs1)               # proper subset
print(fs1.issuperset(fs2))     # fs1 >= fs2
print(fs1.isdisjoint(fs3))     # no common item
```

### frozenset Functions (len() max() min() sum() sorted() in)

```python
num = frozenset({1, 2, 3, 4, 4, 5, 6, 7, 8, 9})
lang = frozenset({"java", "c", "python", "cpp"})
print("Length of frozenset :", len(num))
print("Max of frozenset    :", max(num))
print("Max of frozenset    :", max(lang, key=len))
print("Min of frozenset    :", min(lang))
print("Sum of frozenset    :", sum(num))
print("Sorted              :", sorted(lang))
print("python" in lang)
print("php" not in lang)
```

```python
# convert set <---> frozenset
s = {"a", "b"}
fs = frozenset(s)
s2 = set(fs)
s2.add("c")             # set can change again
print(type(fs), fs)
print(type(s2), s2)
```

```python
# python itself uses frozenset : a constant set in  'in'  is compiled ONCE as a frozenset
import dis
dis.dis("x in {'.jpg', '.png', '.webp'}")
```

## Applications of frozenset (AI , Data , Expert systems)

```python
# 1) Market Basket Analysis (Apriori) : itemsets as dictionary keys
support = {
    frozenset({"bread"}): 0.60,
    frozenset({"bread", "butter"}): 0.32,
    frozenset({"bread", "butter", "milk"}): 0.18,
}
# confidence of the rule  {bread, butter} ---> {milk}
conf = support[frozenset({"butter", "bread", "milk"})] / support[frozenset({"butter", "bread"})]
print(f"confidence = {conf:.2f}")
```

```python
# 2) Undirected graph : edge A-B is the same as edge B-A
distance = {
    frozenset({"Tehran", "Karaj"}): 45,
    frozenset({"Karaj", "Qazvin"}): 105,
}
print(distance[frozenset({"Karaj", "Tehran"})])

# with tuple the order matters
distance_t = {("Tehran", "Karaj"): 45}
print(distance_t.get(("Karaj", "Tehran"), "Not Found"))
```

```python
# 3) Expert system : conditions of rules are unordered and must not change
rules = {
    frozenset({"high_temp", "high_vibration"}): "possible bearing failure",
    frozenset({"high_temp"}): "check the cooling system",
    frozenset({"high_vibration"}): "check the alignment",
}
facts = {"high_temp", "high_vibration", "low_pressure"}

fired = [(cond, diag) for cond, diag in rules.items() if cond <= facts]
fired.sort(key=lambda r: len(r[0]), reverse=True)    # more specific rule first
for cond, diag in fired:
    print(sorted(cond), "--->", diag)
```

```python
# 4) Constants of the program : nobody can change them by mistake
STOP_WORDS_FA = frozenset({"و", "در", "به", "از", "که", "این", "را", "با"})
text = "این مدل در تشخیص تصویر و متن به خوبی کار می‌کند"
tokens = [t for t in text.split() if t not in STOP_WORDS_FA]
print(" | ".join(tokens))
#STOP_WORDS_FA.add("مدل")     # AttributeError ---> protected
```

```python
# 5) Cache key for unordered inputs (lru_cache needs hashable arguments)
from functools import lru_cache

@lru_cache(maxsize=128)
def recommend(user_tags):
    print("   ...computing")          # expensive work
    return sorted(user_tags)

print(recommend(frozenset({"ai", "python"})))
print(recommend(frozenset({"python", "ai"})))   # read from cache , no computing
print(recommend.cache_info())
#recommend({"ai", "python"})     # TypeError: unhashable type: 'set'
```

```python
# 6) State of a search problem (TSP / RL) : (current city , set of visited cities)
visited_states = set()
state1 = ("Tehran", frozenset({"Karaj", "Qazvin"}))
state2 = ("Tehran", frozenset({"Qazvin", "Karaj"}))
visited_states.add(state1)
print(state2 in visited_states)     # same state
```
