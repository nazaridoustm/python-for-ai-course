```python
# str.isdigit()  returns True if all the characters in the string are digits . it returns False if no character is digit in the string
str1= "LotusData2026"
str2= "20261213"
str3="123-45-99"
str4= "IIIV"
str5="\u00B23"   # \u00b2 means unicode character with code U+00B2 
str6="\u00BD"    # it is a numeric character but not a digit
print(str1.isdigit())
print(str2.isdigit())
print(str3.isdigit())
print(str4.isdigit())
print(str5.isdigit())
print(str6.isdigit())

```
